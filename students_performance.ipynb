```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('StudentsPerformance.csv')

print('Dataset shape (row, columns):', df.shape)
print('Colunm name:', df.columns.to_list())

print('Firs 5 rows of the dataset ========================')
print(df.head())

print('Summary statistics of the dataset =================')
print(df.describe())

print('Data types and null info:')
print(df.info())
```

```python
print('Average score by column subject:')
df[["math score", "reading score", "writing score"]].mean()
```

```python
print('Maximum score by column subject:')
df[["math score", "reading score", "writing score"]].max()
```

```python
print('Miniimum score by column subject:')
df[["math score", "reading score", "writing score"]].min()
```

```python
# Correlation
print('Correlation between scores:')
df[["math score", "reading score", "writing score"]].corr()
```

```python
df['total score'] = df[['math score','reading score','writing score']].sum(axis=1)
df['average score'] = df['total score']/3
df['passed'] = np.where(df['average score'] >= 50, 'Yes', 'No')

print('New columns preview (total, average, passed):')
print(df[['total score', 'average score', 'passed']].head())
```

```python
test_prep_group = df.groupby('test preparation course') [['math score','reading score','writing score']].mean()
print('Average score by test preparation course:')
print(test_prep_group)

# Gender-based average
gender_avg = df.groupby('gender') [['math score','reading score','writing score']].mean()
print('Average score by gender:')
print(gender_avg)

# Who passed vs failed
pass_rate = df['passed'].value_counts(normalize=True) * 100
print('Pass rate:\n', pass_rate)
```

```python
# Convert scores to Numpy array
scores = df[['math score','reading score','writing score']].values

# Mean, Std Deviation
print('Numpy mean of scores:\n', np.mean(scores, axis=0))
print('Numpy std of scores:\n', np.std(scores, axis=0))

# Z-score (normalize)
z_scores = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print('Z-score (normalized numbers)')
print(z_scores[:5])

# Average 0
# Negative -
# Positive +  (ya'ni 0dan yuqori bo'lsa yaxshi o'qiyabdi past bo'lsa aksi!)
```

```python
# How many students scored above 90 average?
top_scorers = df[df['average score'] >=90]
print('Number of students who scored above 90 average:', len(top_scorers))

# How many failed?
failed_students = df[df['passed'] == 'No']
print('Number of failed students:', len(failed_students))
```
