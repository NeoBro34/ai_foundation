```python
data_widh_missing = {
    'cars': ['BMW', 'Volvo', None, 'Tesla', 'Ferrari', None, 'Ford'],
    'counts': [3, 7, None, 8, 1, 9, None]
}

df_missing = pd.DataFrame(data_widh_missing)
print(df_missing)
```

```python
df_missing.isnull().sum()
```

```python
df_missing.dropna()
```

```python
df_missing.dropna(how='all') # buyerda qay birida ma'lumot bo'lsa bas olib beradi
```

```python
df_missing.dropna(axis=1)
```

```python
df_missing.dropna(subset=['cars'])
```
